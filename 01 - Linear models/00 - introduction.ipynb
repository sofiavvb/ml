## Local Setup

Creating an isolated environment at the start of a project prevents dependency conflicts and ensures reproducibility.
In Python, there are two common approaches to environment management.

#### Virtual Environment (`venv`)
`venv` is a lightweight, built-in Python module for creating isolated environments.

1. **Create a new environment (for example, `mc886_venv`)**

> ```bash
> python3 -m venv mc886_venv
> ```

This command creates a directory containing a standalone Python interpreter and its own package installation space.
All installed packages remain isolated from the global Python installation.

2. **Activate the environment**

On Linux/macOS:

> ```bash
> source mc886_venv/bin/activate
> ```

On Windows:
> ```bash
> mc886_venv\Scripts\activate
> ```

3. **After activation, any pip install command installs packages locally within this environment**

4. **To deactivate**:

> ```bash
> deactivate
> ```

#### 2. Conda Environment (or miniconda)
[Conda](https://anaconda.org/) is a cross-platform package and environment manager that handles both Python and non-Python dependencies.
It is particularly useful in machine learning workflows that rely on compiled libraries (e.g., CUDA, MKL).

1. **Create a new environment**

> ```bash
> conda create -n mc886_env python=3.12
> ```

2. **Activate the environment**
> ```bash
    > conda activate mc886_env
> ```

3. **After activation, any pip install command installs packages locally within this environment**

4. **To deactivate**:

> ```bash
> conda deactivate
> ```

Conda environments are generally more robust for complex scientific stacks but are heavier than `venv`.

----

### Installing Packages

After activating your environment, install packages using `pip` (or `conda`, if using a Conda environment).
For this hands-on session, only PyTorch and Jupyter are required:

> ```bash
> pip install torch torchvision jupyter
> ```

This installs:
- `torch`: the core PyTorch library
- `torchvision`: common datasets and utilities for computer vision
- `jupyter`: the notebook interface

#### GPU Support:
If your system includes an NVIDIA or AMD GPU, install the appropriate GPU-enabled version of PyTorch.

- For NVIDIA GPUs, install the version compatible with your installed [CUDA](https://developer.nvidia.com/cuda)
- For AMD GPUs, refer to the PyTorch builds compatible with [ROCm](https://rocm.docs.amd.com/en/latest/compatibility/ml-compatibility/pytorch-compatibility.html)

Official installation commands (including specific CUDA/ROCm versions) are provided at https://pytorch.org/get-started/locally/
